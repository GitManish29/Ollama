**Streaming Responses**

```python
from ollama import chat

stream = chat(
    model='gemma3',
    messages=[{'role': 'user', 'content': 'Why is the sky blue?'}],
    stream=True,
)

for chunk in stream:
  print(chunk['message']['content'], end='', flush=True)
```

**Async Client** 

```python
# Cell 1
import asyncio
from ollama import AsyncClient

async def chat():
    client = AsyncClient()
    try:
        messages = [{"role": "user", "content": "Why is the sky blue?"}]
        stream = await client.chat(model="gemma3", messages=messages, stream=True)

        async for part in stream:
            content = (
                part.get("message", {}).get("content")
                or part.get("response")
                or part.get("content")
            )
            if content:
                print(content, end="", flush=True)

        print()  # newline after completion

    except Exception as e:
        print("Error:", e)

    finally:
        aclose = getattr(client, "aclose", None)
        if aclose:
            await aclose()

# Cell 2 (run this cell in same notebook)
await chat()
#Option B — If your environment does not support top-level await, use nest_asyncio
# Install if needed: pip install nest_asyncio
import nest_asyncio
nest_asyncio.apply()

import asyncio
import tracemalloc
import warnings
warnings.filterwarnings("ignore")
from ollama import AsyncClient

async def chat():
    client = AsyncClient()
    try:
        messages = [{"role": "user", "content": "Why is the sky blue?"}]
        stream = await client.chat(model="gemma3", messages=messages, stream=True)

        async for part in stream:
            content = (
                part.get("message", {}).get("content")
                or part.get("response")
                or part.get("content")
            )
            if content:
                print(content, end="", flush=True)

        print()

    except Exception as e:
        print("Error:", e)

    finally:
        aclose = getattr(client, "aclose", None)
        if aclose:
            await aclose()
            tracemalloc.start()

# run via asyncio.run now that nest_asyncio applied
asyncio.run(chat())
```

```python
import litellm
import os
import traceback  # Import traceback for detailed error logging

print("--- Starting litellm direct test (Polished Version) ---")

# --- Setup ---
api_base = "http://127.0.0.1:11434"
model_name = "ollama/gemma3:27b"
os.environ["OLLAMA_API_BASE"] = api_base
print(f"Set OLLAMA_API_BASE to: {api_base}")

# --- 🌟 FIX 1: Enable Verbose Mode ---
# This forces litellm to print its network requests (curl commands)
# This is the most powerful tool for debugging this.
litellm.set_verbose = True
print("Enabled litellm verbose mode. You will now see detailed logs.")

# --- Define Messages ---
# This line just creates the variable. It's not supposed to print.
messages = [
    {"role": "user", "content": "Explain quantum computing in simple terms."}
]

# --- 🌟 FIX 2: Add print statement to fix your confusion ---
# This print statement shows you that the variable was created.
print(f"\nMessages variable is ready: {messages}")

try:
    print("\nAttempting to call litellm.completion()...")
    print("(If it fails, look for a 'curl' command printed above.)")

    # --- 🌟 FIX 3: Add a timeout ---
    # This prevents the cell from hanging forever.
    # If the server is down, it will error out after 30 seconds.
    response = litellm.completion(
        model=model_name, 
        messages=messages,
        timeout=30  # Timeout in seconds
    )
    
    print("\n--- ✅ Call Successful! ---")
    
    # 5. Print the raw response (for debugging)
    print(f"\n[Debug] Raw response object:\n{response}")

    # 6. Print the final content
    final_content = response.choices[0].message.content
    print("\n✅ LiteLLM Response:\n")
    print(final_content)

except Exception as e:
    print("\n--- 🛑 LITELLM TEST FAILED 🛑 ---")
    print(f"An error occurred: {e}")
    print("\nFull Traceback:")
    traceback.print_exc()
    print("\n--- Common Fixes ---")
    print(f"1. Is your Ollama server running? (Check terminal or try opening {api_base} in your browser)")
    print(f"2. Did you pull the model? (Run in terminal: ollama pull gemma3:27b)")

print("\n--- Test finished ---")
```
